# Exercise - 3: Working with Lists and Dictionaries

```python
import csv

students = {}

with open('students.csv','r') as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:
        name = row[0]
        age  = row[1]
        students[name] = age

for name, age in students.items():
    print(name, "is", age, "years old")
```
