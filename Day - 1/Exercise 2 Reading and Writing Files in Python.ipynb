# Exercise - 2: Reading and Writing Files in Python

```python
# read from a file

with open('numbers.csv', 'r') as input_file:

    with open('squares.csv','w') as output_file:

        for line in input_file:

            number = int(line.strip())
            
            square = number ** 2
            
            output_file.write(str(square) + "\n")
```
