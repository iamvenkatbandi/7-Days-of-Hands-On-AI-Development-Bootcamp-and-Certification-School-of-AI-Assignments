# Data Analysis of Survival Rate on Titanic Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('titanic.csv')

data.fillna({'Age': data['Age'].mean()},inplace=True)

print(data['Survived'].value_counts(normalize=True)*100)

print(data.groupby(['Gender','Pclass'])['Survived'].mean())

sns.barplot(x='Gender', y='Survived', data=data)
plt.title('Survival Rate by Gender')
plt.show()
```
