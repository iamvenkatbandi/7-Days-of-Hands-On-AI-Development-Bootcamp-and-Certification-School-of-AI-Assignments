# Finding Median Income and Creating the Histogram of Income Distribution

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('adult.csv')

data['income_num'] = data['income'].map({
    '<=50K': 50000,
    '>50K': 50001
})

print(data.head())
```

```python
median_income = data['income_num'].median()

print('Median Income: ',median_income)
```

```python
plt.figure(figsize=(8,5))

plt.hist(data['income_num'], bins=2)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()
```
