# Pie Chart to Visualize Titanic Passengers Who Survived Vs Who Didn't

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('titanic.csv')

print(data.head())
```

```python
print('0 - Not Survived\n1 - Not Survived')

data['Survived'].value_counts().plot(kind='pie')
plt.title('Survived vs Not Survived')
plt.show()
```
