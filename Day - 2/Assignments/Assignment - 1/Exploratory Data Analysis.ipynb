# Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('cybersecurity_threat_dataset.csv')

print(data.head())
```

```python
print(data.isnull().sum())
```

```python
data.fillna({'tags':'N/A'},inplace=True)
```

```python
print(data.isnull().sum())
```

```python
print(data.shape)
```

```python
print(data.info())
```

```python
import matplotlib.pyplot as plt

plt.plot(data['threat_type'],data['days_active'])
plt.title('Number of Days Threat Active')
plt.xlabel('Threat Type')
plt.ylabel('Number of Days Active')
plt.show()
```

```python
data['threat_type'].value_counts().plot(kind='bar')
plt.title('Threat Type')
plt.show()
```

```python
data['malware_family'].value_counts().plot(kind='bar')
plt.title('Malware Family')
plt.show()
```

```python
data['threat_type'].value_counts().plot(kind='pie')
plt.title('Threat Type')
plt.show()
```

```python
sns.pairplot(data[['threat_type','malware_family','days_active']])
plt.show()
```
